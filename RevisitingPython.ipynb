```python
#Importing the library needed to plot graphs
import matplotlib.pyplot as plt
%matplotlib inline  

class Circle(object): #Will create a class
    
    # Constructor
    def __init__(self, radius=3, color='blue'):
        self.radius = radius
        self.color = color 
    
    # Method
    def add_radius(self, r):
        self.radius = self.radius + r
        return(self.radius)
    
    # Method
    def drawCircle(self):
        plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))
        plt.axis('scaled')
        plt.show()  
#We can then create different objects in that class
RedCircle = Circle(10, 'red') #Number 10 will be stored as radius, red as color. 
RedCircle.radius #Will print what was saved as the characteristic of the object which can
                #can be changes using RedCircle.radius = 1
RedCircle.drawCircle() #Will call the method, which is a function inside the code of class in this case
```

```python
# Creating a Program to Count words in a String using Dictionary
def freq(string,passedkey):

    #A list variable is declared and initialized to an empty list.
    words = []

    #Breaking the string into list of words
    words = string.split() # or string.lower().split()

    #Declare a dictionary
    Dict = {}

    #Use for loop to iterate words and values to the dictionary
    for key in words:
        if(key == passedkey):
            Dict[key] = words.count(key)   
    #step5: Print the dictionary
    print("Total Count:",Dict)

#step6: Call function and pass string in it
freq("Mary had a little lamb Little lamb, little lamb Mary had a little lamb.Its fleece was white as snow And everywhere that Mary went Mary went, Mary went Everywhere that Mary went The lamb was sure to go","little")
```
